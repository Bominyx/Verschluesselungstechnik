```python
def nilpotent_numbers(n):
    numbers = []
    for g in range(0, n):
        for N in range(1, n + 1):
            #print(f"{g}^{N} % {n} = {g**N % n}")
            if g**N % n == 0:
                numbers.append(g)
                break
    return numbers
```

```python
nilpotent_numbers(24)
```

```python
nilpotent_numbers(120)
```

```python
for n in range(2, 51):
    print(f"{n}:{nilpotent_numbers(n)}")
```

```python
# O(n)
def power_mod(x, e, n):
    result = 1
    for _ in range(e):
        result = (result * x) % n
    return result
```

```python
# O(log(n))
# max 2*k Multiplikationen
def fast_power_mod(x, e, n):
    if e == 1:
        return x
    if e % 2 == 0:
        return (fast_power_mod(x, e//2, n)**2) % n
    else:
        return (x*fast_power_mod(x, e-1, n)) % n
```

```python
fast_power_mod(3, 13, 100)
```

```python
fast_power_mod(3, 10**18, 1000003)
```
